# 01 Data Understanding

Inspect dataset dimensions, columns, missingness, duplicates, and measure definitions.

```python
from pathlib import Path
import pandas as pd
DATA = Path('../data/raw/heart_disease_data.csv')
df = pd.read_csv(DATA)
print('Shape:', df.shape)
display(df.head())
```

```python
display(pd.DataFrame({'dtype': df.dtypes.astype(str), 'missing': df.isna().sum(), 'missing_pct': (df.isna().mean()*100).round(2)}))
```

```python
print('Exact duplicate rows:', df.duplicated().sum())
print('Years:', df['Year'].min(), '-', df['Year'].max())
print('Topics:', sorted(df['Topic'].dropna().unique()))
```
