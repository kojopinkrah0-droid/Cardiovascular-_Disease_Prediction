# 03 Exploratory Data Analysis

Explore yearly, topic, location, and breakout patterns with charts.

```python
from pathlib import Path
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt
from src.data_cleaning import load_dataset, clean_dataset
df = clean_dataset(load_dataset('../data/raw/heart_disease_data.csv'))
```

```python
display(df.groupby('topic')['data_value'].agg(['mean','median','count']).sort_values('mean', ascending=False))
```

```python
df.groupby('year')['data_value'].mean().plot(marker='o', figsize=(9,5), title='Mean reported value by year'); plt.ylabel('Data_Value'); plt.grid(alpha=.25); plt.show()
```

```python
df.groupby('topic')['data_value'].mean().sort_values().plot(kind='barh', figsize=(9,5), title='Mean reported value by topic'); plt.xlabel('Mean Data_Value'); plt.show()
```

```python
df.boxplot(column='data_value', by='break_out_category', rot=20, figsize=(9,5)); plt.suptitle(''); plt.title('Reported values by breakout category'); plt.show()
```
